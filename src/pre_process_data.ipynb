```python
import pandas as pd
import os

# Ensure folder exists
os.makedirs("../data", exist_ok=True)

# NASA GISS global temperature data URL
url = "https://data.giss.nasa.gov/gistemp/tabledata_v4/GLB.Ts+dSST.csv"

# Read CSV from web, skip metadata row
df = pd.read_csv(url, skiprows=1)

# Keep 'Year' and 'J-D' (Jan-Dec average) columns
df = df[['Year', 'J-D']]

# Rename columns
df.columns = ['year', 'temperature_increase']

# Convert to numeric, coerce errors (e.g., '***') to NaN
df['temperature_increase'] = pd.to_numeric(df['temperature_increase'], errors='coerce')

# Drop rows with NaN values
df = df.dropna()

# Convert to absolute value
df['temperature_increase'] = df['temperature_increase'].abs()

# Save CSV
df.to_csv("../data/global_temperature_increase.csv", index=False)

print("CSV saved as '../data/global_temperature_increase.csv'")
```
